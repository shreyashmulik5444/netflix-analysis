# feature engineering

### fe for 'date_added' column

```python
df['date_added'] = pd.to_datetime(df['date_added'],errors='coerce') #errors='coerce': Valid date = converted properly , Invalid date = replaced with NaT

df['year_added'] = df['date_added'].dt.year # '.dt is a datetime accessor in pandas'
df['month_added'] = df['date_added'].dt.month
df['day_added'] = df['date_added'].dt.day
```
